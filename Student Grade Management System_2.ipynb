```python
Students=[]
```

```python
def StudentStatus(average):
    if average >= 85:
        Status="Excellent"
    elif average >=75:
       Status="Very Good"
    elif average >=60:
        Status="Good"
    elif average >=50:
        Status="Passed"
    else:
        Status="Failed"
    return Status
```

```python
#Add Student-->option1
def add_student():
    subjectWithGrade = {}
    
    #studentinformation
    student_name = input("Enter student name:")
    try:
        student_age = int(input("Enter age:"))
    except ValueError:
        print("Invalid age!")
        student_age = int(input("Enter age:"))
    student_department = input("Enter department:")
    
    try:
        numberOfSubject = int(input("How many subjects?"))
    except ValueError:
        print("Invalid number!")
        numberOfSubject = int(input("How many subjects?"))
    for n in range(1,numberOfSubject+1):
        subjectName = input(f"Enter subject name {n}:")

        #subject Grade number between 0 to 100
        flag = True
        while flag:
            try:
                subjectGrade = float(input(f"Enter grade for subject {n}:"))
                if subjectGrade < 0 or subjectGrade > 100:
                        print("Invalid Grade")
                else:
                    subjectWithGrade[subjectName]=subjectGrade
                    flag = False
            except ValueError:
                    print("Invalid Grade!Please enter a number")
    #calculate average
    average = sum(subjectWithGrade.values())/len(subjectWithGrade)

    #Student Status
    status= StudentStatus(average)
    
    #addition
    Students.append({"name":student_name, "age":student_age, "department":student_department, "average":average, "status":status, "subjectsAndGrade":subjectWithGrade})
```

```python
#Show Students-->option2
def show_students():
    if not Students:
        print("NO STUDENT FOUND!!!")
        return
    print("Students:")
    for student in Students:
        print("--------------------")
        print(f"Name: {student['name']}")
        print(f"Age: {student['age']}")
        print(f"Department: {student['department']}")
        for subject,grade in student['subjectsAndGrade'].items():
            print(f"{subject}: {grade}")
        print(f"Average: {student['average']}")
        print(f"Status: {student['status']}")
```

```python
#Search Student-->option3
def Search_Student():
    student_search_name = input("Enter The name: ")
    flag = False
    for student in Students:
        if student_search_name == student["name"]:
            print(f"Student Found! \n name: {student['name']} \n age: {student['age']} \n department: {student['department']}")
            print(f"Subjects & Grades: \n {student['subjectsAndGrade']}")
            print(f"average: {student['average']}")
            print(f"Status: {student['status']}")
            flag = True
            break
    if not flag: 
        print("Student not found.")
```

```python
#calculate average-->option4
def calculate_average():
    if not Students:
        print("NO STUDENT FOUND!!!")
        return
    flag = False
    while not flag:
        student_to_average = input("Enter The name:")
        for s in Students:
            if student_to_average == s["name"]:
                print(f"avarage: {s['average']} \n Status: {s['status']}")
                flag =True
                break
        if not flag:
            print("Student not found.")
        
```

```python
#show statistics-->option5
def show_statistics():
    if not Students:
        print("NO STUDENT FOUND!!!")
        return
    NumberOFExcellent=0
    NumberOFVeryGood=0
    NumberOFGood=0
    NumberOFPassed=0
    NumberOfFailed=0

    print("STATISTICS:")
    #Total Students
    print(f"Total Students: {len(Students)}")
    
    ##Highest and Lowest Average
    averages = [s["average"] for s in Students]
    statuses = [s["status"] for s in Students]
    print(f"Highest Average: {max(averages)}")
    print(f"Lowest Average: {min(averages)}")
    print(f"Class Average: {sum(averages)/len(averages)}")

    for st in statuses:
        if st=="Excellent":
            NumberOFExcellent+=1
        elif st=="Very Good":
            NumberOFVeryGood+=1
        elif st=="Good":
            NumberOFGood +=1
        elif st=="Passed":
            NumberOFPassed+=1
        elif st=="Failed":
            NumberOfFailed+=1
    print(f"Number of Excellent: {NumberOFExcellent} \nNumber of Very Good: {NumberOFVeryGood}")
    print(f"Number of Good: {NumberOFGood}\nNumber of Passed: {NumberOFPassed} \nNumber of Failed: {NumberOfFailed}")
```

```python
running = True
while running:
    print(f"STUDENT GRADE MANAGEMENT: \n 1. Add Student \n 2. Show Students \n 3. Search Student \n 4. Calculate Average \n 5. Show Statistics \n 6. Exit")
    option = int(input("Choose an option: "))
    if option==1:
        add_student()
    elif option==2:
        show_students()
    elif option==3:
        Search_Student()
    elif option==4:
        calculate_average()
    elif option==5:
        show_statistics()
    elif option==6:
        print("GOOD BYE!")
        running=False
    else:
        print("Invalid option!")
```
